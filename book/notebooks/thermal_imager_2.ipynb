# Custom dimension iteration

Adds the water-vapour column to the surface temperature and emissivity sweep of {doc}`thermal_imager` by calling `run()` in a loop with a different `params` value each time and concatenating the results.

## Approach

- Surface temperature and albedo (1 − emissivity) are dataset variables, as in {doc}`thermal_imager`.
- The water-vapour column (mm) is set per run with `params={'mol_modify H2O': value}`. The results are concatenated along a new `h2o_column` dimension.
- The same loop sweeps any other uvspec option: `params` passes key–value pairs to the libRadtran input file ({doc}`/guide/parameters`).

```python
import pyradtran
from pyradtran import Var
import matplotlib.pyplot as plt
import numpy as np
import xarray as xr
from pathlib import Path
import pandas as pd
import logging

logging.getLogger('pyradtran').setLevel(logging.CRITICAL)

config_path = Path('config/velox.yaml')

ds = xr.Dataset(
    coords={
        'time' : '2025-04-04T23:00:00',
        'latitude' : [78.925],
        'longitude' : [12.3747329],
        'altitude' : ('altitude', [0, 10, 100]),
    },
    data_vars={
        'surface_temperature' : [253.15, 263.15, 273.15, 283.15, 293.15],
        'albedo' : [0, 0.25, 0.5, 0.75, 1],
    }
)
ds_out = []
h2o_values = np.arange(0, 40, 5)

for h2o in h2o_values:
    ds_sim = ds.pyradtran.run(
        config_path=config_path,
        return_dataset=True,
        save_to_file=False,
        params={
            'sur_temperature': Var('surface_temperature'),
            'albedo': Var('albedo'),
            'output_quantity' : 'brightness',
            'mol_modify H2O': float(h2o),        # water-vapour column in mm
            'execution.max_workers': 8,
        },
        show_progress=False,
    )
    ds_out.append(ds_sim)

ds_out = xr.concat(ds_out, dim='h2o_column')
ds_out = ds_out.assign_coords(h2o_column=h2o_values)
print("\nSimulation complete!")
```

## Results: atmospheric effect on brightness temperature

Each panel shows, for one water-vapour column, the upwelling brightness temperature (7.7–12 µm, US standard atmosphere) at 100 km minus that at the surface. Positive values mean the atmosphere adds more emission than it absorbs, negative values the opposite. Cold or weakly emitting surfaces look warmer from above, warm and strongly emitting surfaces look colder, and both effects are larger for moist columns than for dry ones. At emissivity 0 the surface only reflects the downwelling radiation, so the difference is positive for every surface temperature.

```python
fig, ax = plt.subplots(2, 4 , figsize=(14, 6))
ax = ax.flatten()

for i, h2o_val in enumerate(h2o_values):
    ds_subset = ds_out.sel(h2o_column=h2o_val)
    delta_atm = ds_subset.sel(altitude=100) - ds_subset.sel(altitude=0)
    delta_atm['emissivity'] = 1 - delta_atm['albedo']
    delta_atm = delta_atm.swap_dims({'albedo': 'emissivity'})

    delta_atm.eup.plot(ax=ax[i], vmin=-10, vmax=10, cmap='RdBu_r', x='emissivity',
                       cbar_kwargs={'label': 'BT(100 km) − BT(0 km) (K)'})
    ax[i].set_title(f"H₂O = {h2o_val:.0f} mm")
    ax[i].set_ylabel('Surface temperature (K)')
plt.tight_layout()
```
