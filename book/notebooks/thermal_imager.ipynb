# Brightness temperature

Brightness temperatures in the 7.7–12 µm band for five surface temperatures and five albedo values, at 0, 10 and 100 km. The run uses `config/velox.yaml`; the uvspec option `output_quantity brightness` converts the output to brightness temperature.

```python
import pyradtran  # Registers the .pyradtran xarray accessor
from pyradtran import Var
import matplotlib.pyplot as plt
import numpy as np
import xarray as xr
from pathlib import Path
import pandas as pd
import logging

logging.getLogger('pyradtran').setLevel(logging.WARNING)  # warnings only

config_path = Path('config/velox.yaml')

ds = xr.Dataset(
    coords={
        'time' : '2025-04-04T23:00:00',
        'latitude' : [78.925],
        'longitude' : [12.3747329],
        'altitude' : ('altitude', [0, 10, 100]),
    },
    data_vars={
        'surface_temperature' : [253.15, 263.15, 273.15, 283.15, 293.15],
        'albedo' : [0, 0.25, 0.5, 0.75, 1],
    }
)

ds_sim = ds.pyradtran.run(
    config_path=config_path,
    return_dataset=True,
    save_to_file=True,
    params={
        'sur_temperature': Var('surface_temperature'),
        'albedo': Var('albedo'),
        'output_quantity' : 'brightness',
    },
    show_progress=False,
)

print("\nSimulation complete!")
print(f"Dataset variables: {list(ds_sim.data_vars.keys())}")
print(f"Dataset coordinates: {list(ds_sim.coords.keys())}")
print(f"Dataset shape: {dict(ds_sim.sizes)}")
```

```python
fig, ax = plt.subplots(1, 3, figsize=(12, 4), sharey=True)

ds_sim.eup.sel(altitude=0).squeeze().plot(ax=ax[0], vmin=230, vmax=300, cbar_kwargs={'label': 'Brightness temperature (K)'})
ds_sim.eup.sel(altitude=10).squeeze().plot(ax=ax[1], vmin=230, vmax=300, cbar_kwargs={'label': 'Brightness temperature (K)'})
ds_sim.eup.sel(altitude=100).squeeze().plot(ax=ax[2], vmin=230, vmax=300, cbar_kwargs={'label': 'Brightness temperature (K)'})

for a in ax:
    a.set_xlabel('Albedo')
    a.set_ylabel('Surface temperature (K)')
ax[0].set_title('Surface (0 km)')
ax[1].set_title('10 km')
ax[2].set_title('100 km')

fig.suptitle(r'Upward band brightness temperature (7.7–12 µm) vs surface temperature and albedo ($\alpha = 1-\varepsilon$)')
plt.tight_layout()
plt.show()
```

## Emissivity

libRadtran sets the surface emissivity through the albedo:

$$
\varepsilon = 1 - \alpha
$$

At 0 km the upward brightness temperature is the surface temperature for $\alpha = 0$ (a black body: 293.1 K for 293.15 K). As the albedo grows, less is emitted and more of the cold sky is reflected; at $\alpha = 1$ the surface emits nothing and all five rows show the reflected sky, 202 K. At 10 km and 100 km the atmosphere (here the US standard atmosphere, `afglus`) adds its own emission and absorbs part of the surface signal, so warm surfaces appear 3–8 K colder and the reflecting surfaces slightly warmer.
