# Realistic simulation of VELOX thermal infrared channels

A first-order forward model for the six channels of the VELOX thermal infrared imager on the HALO research aircraft (Schäfer et al., 2022, *Atmos. Meas. Tech.*, 15, 1491–1509, https://doi.org/10.5194/amt-15-1491-2022), for Arctic sea-ice scenes of the HALO-(AC)³ campaign. It shows how the channels respond to surface type, skin temperature, emissivity and water-vapour column, applies first-order instrument corrections (germanium window, channel Tmax, NEDT) and compares the result with broadband VELOX observations from RF04. The RF04 file is not public.

```{warning}
This is a first-order VELOX channel simulator, not a validated forward model. Spectral response functions are top-hat filters built from published wavelength bounds and peak transmissions, the surface emissivities are illustrative, and there is no calibration chain or retrieval.
```

## 1. Why simulate VELOX channels?

A broadband thermal simulation (7.7–12.0 µm) gives one brightness temperature. VELOX observes the scene in six channels, and the channel differences carry independent information:

| Channel pair | Physical signal |
|---|---|
| CH3 (10.74 µm) − CH6 (12.00 µm) | Classic split-window: H₂O absorption, skin-T, emissivity |
| CH5 (11.66 µm) − CH6 (12.00 µm) | Narrower split-window |
| Broadband − narrowband | Atmospheric weighting function contrast |
| CH2 (8.65 µm) vs window channels | Cloud-top detection, cloud phase proxy |

## 2. Instrument model

VELOX (**V**ideo airborn**E** **L**ongwave **O**bservations within si**X** channels) looks at nadir from HALO (Gulfstream G-550, typical survey altitude 8–12 km).

* Sensor format: 640 × 512 px (post-processing uses 635 × 507 px for cross-filter comparability)
* Field of view: 35.5° × 28.7°
* Thermal range: 7.7–12.0 µm

The cells below define the six filters, with spectral response functions (SRFs) approximated as top-hat filters. Replace `load_srf_or_tophat()` to use measured SRF files.

````python
def make_velox_channels():
    """Return a DataFrame with VELOX channel definitions.

    Wavelength bounds are derived from Schäfer et al. (2022) Table 1
    (centre ± half-width). SRFs are *approximated* as top-hat filters;
    replace with measured SRF data when available.

    Returns
    -------
    pd.DataFrame
        Columns: channel, wl_min [nm], wl_max [nm], center_um, halfwidth_um,
        Tmax, label, ch_type.
    """
    rows = [
        # ch  wl_min  wl_max  center  hw     Tmax  label                type
        (1,  7700,  12000,  None,   None,  0.94, "CH1 Broadband",     "broadband"),
        (2,  8100,   9200,  8.65,   0.55,  0.89, "CH2 8.65 µm",       "narrowband"),
        (3, 10350,  11130, 10.74,   0.39,  0.83, "CH3 10.74 µm",      "narrowband"),
        (4,  7700,  12000,  None,   None,  0.94, "CH4 Broadband",     "broadband"),
        (5, 10850,  12470, 11.66,   0.81,  0.88, "CH5 11.66 µm",      "narrowband"),
        (6, 11500,  12500, 12.00,   0.50,  0.92, "CH6 12.00 µm",      "narrowband"),
    ]
    cols = ["channel", "wl_min", "wl_max", "center_um", "halfwidth_um",
            "Tmax", "label", "ch_type"]
    df = pd.DataFrame(rows, columns=cols)
    # Fill broadband centre as midpoint for reference only
    mask = df["center_um"].isna()
    df.loc[mask, "center_um"] = (df.loc[mask, "wl_min"] + df.loc[mask, "wl_max"]) / 2e3
    return df


channels = make_velox_channels()
print(channels[["channel", "wl_min", "wl_max", "center_um", "Tmax", "label"]].to_string(index=False))
````

````python
def plot_channel_response(channels_df, ax=None):
    """Plot top-hat channel response functions for all VELOX channels."""
    if ax is None:
        fig, ax = plt.subplots(figsize=(10, 3.5))
    else:
        fig = ax.get_figure()

    colors = plt.cm.tab10(np.linspace(0, 0.9, len(channels_df)))
    wl_grid = np.linspace(7600, 12700, 1000)  # nm

    for (_, ch), color in zip(channels_df.iterrows(), colors):
        response = np.where(
            (wl_grid >= ch["wl_min"]) & (wl_grid <= ch["wl_max"]),
            ch["Tmax"], 0.0
        )
        ax.fill_between(wl_grid / 1e3, 0, response, alpha=0.35, color=color)
        ax.step(wl_grid / 1e3, response, color=color, lw=1.5,
                label=f"{ch['label']} (T$_{{max}}$={ch['Tmax']:.2f})")
        # Annotate centre wavelength
        cx = ch["center_um"]
        ax.text(cx, ch["Tmax"] + 0.015, f"CH{int(ch['channel'])}",
                ha="center", va="bottom", fontsize=8, color=color)

    ax.set_xlabel("Wavelength (µm)")
    ax.set_ylabel("Transmittance (top-hat)")
    ax.set_xlim(7.5, 12.8)
    ax.set_ylim(0, 1.15)
    ax.legend(fontsize=8, ncol=2, loc="upper right")
    ax.set_title("VELOX channel top-hat filters (approximated SRF)")
    fig.tight_layout()
    return fig, ax


fig_ch, _ = plot_channel_response(channels)
plt.show()
````

````python
def load_srf_or_tophat(channel_row, wl_grid_nm):
    """Return normalised SRF for a VELOX channel on *wl_grid_nm*.

    Currently returns a top-hat based on the published wavelength range
    and Tmax.  **Replace this function** to load measured SRF curves from
    a CSV file (columns: wavelength_nm, transmittance) when available.

    Parameters
    ----------
    channel_row : pd.Series
        One row from make_velox_channels().
    wl_grid_nm : array-like
        Wavelength grid in nm.

    Returns
    -------
    np.ndarray
        Normalised SRF (0-1) on wl_grid_nm.

    Example for real SRF
    --------------------
    # srf_df = pd.read_csv(f'srf_ch{int(channel_row.channel)}.csv')
    # return np.interp(wl_grid_nm, srf_df['wavelength_nm'], srf_df['transmittance'],
    #                  left=0, right=0)
    """
    wl = np.asarray(wl_grid_nm)
    return np.where(
        (wl >= channel_row["wl_min"]) & (wl <= channel_row["wl_max"]),
        channel_row["Tmax"], 0.0
    )
````

## 3. Scene definition: Arctic sea-ice case (HALO-AC³ RF04 context)

The scene is loosely based on HALO-(AC)³ research flight 04 (2022-03-14), a transit and survey flight over the Svalbard sea-ice region.

**Atmosphere**: AFGL subarctic winter profile (`afglsw`), with the water-vapour column set by `mol_modify H2O`. A US standard atmosphere (288 K at the surface) would put warm, moist air above a 253 K surface and make the channels warmer than the surface. For realistic profiles use ERA5 or radiosonde data ({doc}`/guide/atmosphere`).

**Surface types** (all approximate):

| Type | Skin temp | Emissivity | Note |
|---|---|---|---|
| Open water / lead | 271.35 K | 0.98 | Near freezing point of saline water |
| Snow-covered sea ice | 253 K | 0.99 | Cold Arctic sea ice |
| Thin/bare ice | 265 K | 0.95 | Transitional, more variable emissivity |

These emissivities are rough values. VELOX retrievals need spectrally resolved emissivity spectra (e.g. from laboratory databases or the ASTER spectral library).

## 4. Radiative transfer setup

Each channel is one uvspec run integrated over its top-hat wavelength window:

* `source thermal` and `output_process integrate`: band-integrated signal
* `uu` with `umu 1`: upwelling radiance seen by a nadir-looking sensor (what VELOX measures), next to the hemispheric irradiance `eup`
* `output_quantity brightness`: band-integrated radiance converted to brightness temperature in kelvin ({doc}`/guide/outputs`)
* `reptran medium`: molecular absorption at moderate resolution
* per-channel wavelength range via `params={'wavelength': 'wl_min wl_max'}`
* surface emissivity entered as `albedo = 1 − ε`

`run_velox_channel_scene` loops over the six channels and returns one dataset with dimensions `(channel, surface_type, altitude)`. Because the filters are top-hats, the band-integrated brightness temperature does not depend on $T_{\max}$.

````python
# ── Config ─────────────────────────────────────────────────────────────────────
CONFIG_PATH = Path("config/velox_realistic.yaml")
assert CONFIG_PATH.exists(), f"Config not found: {CONFIG_PATH.resolve()}"

# Set RUN_EXPENSIVE = True to run the full H₂O sweep (8 values).
# The default run is: 6 channels × 3 surface types × 2 H₂O values = 36 simulations.
RUN_EXPENSIVE = False
````

````python
def run_velox_channel_scene(
    channels_df,
    surface_types,
    config_path,
    h2o_mm=5.0,
    aircraft_altitude_km=10.0,
    lat=LAT,
    lon=LON,
    base_time=BASE_TIME,
    mol_abs_param=None,
    extra_params=None,
):
    """Run a per-channel VELOX forward simulation over multiple surface types.

    For each VELOX channel the wavelength range is overridden at runtime via
    ``params={'wavelength': 'wl_min wl_max'}``.  All other
    settings come from *config_path* (velox_realistic.yaml).

    Parameters
    ----------
    channels_df : pd.DataFrame
        VELOX channel table from ``make_velox_channels()``.
    surface_types : dict
        ``{name: {'T_sfc': float, 'emissivity': float}}``.
    config_path : Path
        YAML config file path.
    h2o_mm : float
        Precipitable water column in mm (``mol_modify H2O``).
    aircraft_altitude_km : float
        Output altitude (zout) in km — corresponds to aircraft level.
    lat, lon : float
        Scene location (°N, °E).
    base_time : str
        ISO-format UTC timestamp.
    mol_abs_param : str, optional
        Override ``mol_abs_param`` (e.g. 'reptran coarse' for speed).
    extra_params : dict, optional
        Further ``params`` entries, e.g. a cloud (``wc_file``).

    Returns
    -------
    xr.Dataset
        Dimensions ``(channel, surface_type, altitude)``.
        Key variable: ``uu`` — nadir brightness temperature in K.
    """
    stype_names = list(surface_types.keys())
    T_sfcs   = [surface_types[s]["T_sfc"]                  for s in stype_names]
    albedos  = [1.0 - surface_types[s]["emissivity"]       for s in stype_names]

    # One time step per surface type so (T_sfc, albedo) pairs are preserved
    times = pd.date_range(base_time, periods=len(stype_names), freq="s")

    ds_input = xr.Dataset(
        coords={
            "time":      times,
            "latitude":  ("time", [lat] * len(stype_names)),
            "longitude": ("time", [lon] * len(stype_names)),
            "altitude":  ("altitude", [aircraft_altitude_km]),
        },
        data_vars={
            "surface_temperature": ("time", T_sfcs),
            "albedo":              ("time", albedos),
        },
    )

    # Build base overrides
    base_overrides = {
        "output_quantity": "brightness",
        "mol_modify H2O": float(h2o_mm),   # mm
    }
    if mol_abs_param is not None:
        base_overrides["mol_abs_param"] = mol_abs_param
    base_overrides.update(extra_params or {})

    channel_results = []
    for _, ch in channels_df.iterrows():
        overrides = {**base_overrides,
                     "wavelength": f"{int(ch['wl_min'])} {int(ch['wl_max'])}"}

        ds_ch = ds_input.pyradtran.run(
            config_path=config_path,
            return_dataset=True,
            save_to_file=False,
            params={'sur_temperature': Var('surface_temperature'), 'albedo': Var('albedo'), **overrides},
            show_progress=False,
        )

        # Label the time dimension with surface type names
        ds_ch = ds_ch.assign_coords(time=stype_names).rename({"time": "surface_type"})
        channel_results.append(ds_ch)

    ds_all = xr.concat(channel_results, dim="channel")
    ds_all = ds_all.assign_coords(channel=channels_df["channel"].values)

    # Attach channel metadata as coordinate
    ds_all = ds_all.assign_coords(
        channel_label=("channel", channels_df["label"].values),
        Tmax=("channel", channels_df["Tmax"].values),
        wl_min_nm=("channel", channels_df["wl_min"].values),
        wl_max_nm=("channel", channels_df["wl_max"].values),
    )
    ds_all.attrs.update({
        "h2o_mm":          h2o_mm,
        "aircraft_alt_km": aircraft_altitude_km,
        "lat": lat, "lon": lon,
        "srf_type": "top-hat (band limits only; Tmax not applied)",
    })
    return ds_all
````

````python
# ── Default run: dry Arctic atmosphere (H₂O = 5 mm) ──────────────────────────
print("Running 6 channels × 3 surface types (dry Arctic, H₂O = 5 mm)...")
ds_dry = run_velox_channel_scene(
    channels, SURFACE_TYPES, CONFIG_PATH, h2o_mm=H2O_DRY_MM
)

# ── Moist Arctic (H₂O = 20 mm) for split-window sensitivity ──────────────────
print("Running 6 channels × 3 surface types (moist Arctic, H₂O = 20 mm)...")
ds_moist = run_velox_channel_scene(
    channels, SURFACE_TYPES, CONFIG_PATH, h2o_mm=H2O_MOIST_MM
)

print("Done.")
````

## 5. Instrument effects

The simulation gives the scene brightness temperature. The raw signal of the camera also contains the filter and window effects below. A blackbody calibration through the same optics removes them, so calibrated VELOX brightness temperatures are compared with the *uncorrected* simulation (section 7). The corrected values show how large the effects are that the calibration has to remove.

### 5a. Filter peak transmittance
A real filter transmits only $T_{\max} < 1$ of the scene radiance:
$$L_{\mathrm{eff}} = T_{\max} \cdot L_{\mathrm{scene}} + (1 - T_{\max}) \cdot L_{\mathrm{self}}$$
where $L_{\mathrm{self}}$ is the self-emission of the filter, negligible for a cold filter (80 K assumed here). The code evaluates Planck's law at the channel centre, which is rough for the broadband channels.

### 5b. Germanium window
VELOX looks through a germanium window with transmittance $\tau_{\mathrm{win}} \approx 0.9395$:
$$L_{\mathrm{meas}} = \tau_{\mathrm{win}} \cdot L_{\mathrm{scene}} + (1 - \tau_{\mathrm{win}}) \cdot B(\lambda_{\mathrm{eff}}, T_{\mathrm{win}})$$
$T_{\mathrm{win}}$ is the physical temperature of the window (293 K assumed), so the window adds a warm offset. $\tau_{\mathrm{win}}$ depends on wavelength; a single constant is used here.

### 5c. NEDT noise
`add_nedt_noise` adds Gaussian noise with an assumed NEDT of 0.04 K.

````python
# Physical constants
_h  = 6.62607015e-34  # J s
_c  = 2.99792458e8    # m/s
_kb = 1.380649e-23    # J/K


def planck_radiance(T_K, wl_um):
    """Spectral radiance [W m⁻² sr⁻¹ µm⁻¹] from Planck's law.

    Parameters
    ----------
    T_K : float or array
        Temperature in Kelvin.
    wl_um : float or array
        Wavelength in µm.
    """
    wl_m = np.asarray(wl_um) * 1e-6
    T    = np.asarray(T_K)
    exponent = (_h * _c) / (wl_m * _kb * T)
    return (2 * _h * _c**2 / wl_m**5) / (np.exp(exponent) - 1) * 1e-6  # µm⁻¹


def brightness_temperature(L_wm2sr_um, wl_um):
    """Invert Planck's law: radiance [W m⁻² sr⁻¹ µm⁻¹] → brightness temperature [K]."""
    wl_m = np.asarray(wl_um) * 1e-6
    L    = np.asarray(L_wm2sr_um)
    return (_h * _c) / (wl_m * _kb * np.log(1 + (2 * _h * _c**2) / (wl_m**5 * L * 1e6)))
````

````python
TAU_WINDOW   = 0.9395   # Germanium window transmittance (constant, first-order)
T_WINDOW_K   = 293.15   # Window physical temperature (cabin ~20°C)
NEDT_K       = 0.04     # Noise-equivalent temperature difference (K)


def apply_window_correction(BT_K, wl_eff_um, tau=TAU_WINDOW, T_win=T_WINDOW_K):
    """Apply the germanium window transmission to a scene BT (raw-signal effect).

    Works in radiance space at the channel's effective wavelength.

    Lmeas = tau * L_scene + (1-tau) * B(wl_eff, T_win)
    BT_corrected = B_inv(Lmeas, wl_eff)

    Parameters
    ----------
    BT_K : float or array
        Scene brightness temperature before window (K).
    wl_eff_um : float
        Channel effective wavelength (µm).
    tau : float
        Window transmittance.
    T_win : float
        Window physical temperature (K).

    Returns
    -------
    np.ndarray
        Corrected brightness temperature (K).

    Warnings
    --------
    tau is wavelength-independent here — a first-order approximation only.
    """
    L_scene = planck_radiance(BT_K, wl_eff_um)
    L_win   = planck_radiance(T_win, wl_eff_um)
    L_meas  = tau * L_scene + (1 - tau) * L_win
    return brightness_temperature(L_meas, wl_eff_um)


def apply_tmax_correction(BT_K, Tmax, wl_eff_um, T_filter_K=80.0):
    """Apply the peak filter transmittance to a scene BT (raw-signal effect).

    Leff = Tmax * L_scene + (1-Tmax) * B(wl_eff, T_filter)
    For a cold filter (T_filter ≈ 80 K) the self-emission term is
    negligible; for a warm filter use its actual temperature.

    Warnings
    --------
    Linearisation is approximate, especially for broadband channels.
    """
    L_scene = planck_radiance(BT_K, wl_eff_um)
    L_self  = planck_radiance(T_filter_K, wl_eff_um)
    L_eff   = Tmax * L_scene + (1 - Tmax) * L_self
    return brightness_temperature(L_eff, wl_eff_um)


def add_nedt_noise(BT_K, nedt_k=NEDT_K, seed=42):
    """Add Gaussian NEDT noise to a brightness temperature array."""
    rng = np.random.default_rng(seed)
    return np.asarray(BT_K) + rng.normal(0.0, nedt_k, size=np.asarray(BT_K).shape)
````

````python
# ── Apply all corrections to the dry-Arctic run ───────────────────────────────
BT_raw     = ds_dry.uu.isel(altitude=0).values   # shape (channel, surface_type)
BT_corrected = np.empty_like(BT_raw)

for i, (_, ch) in enumerate(channels.iterrows()):
    wl_eff = ch["center_um"]
    bt = BT_raw[i]
    bt = apply_tmax_correction(bt, ch["Tmax"], wl_eff)
    bt = apply_window_correction(bt, wl_eff)
    BT_corrected[i] = bt

# Build a corrected xarray DataArray
BT_corrected_da = xr.DataArray(
    BT_corrected,
    dims=["channel", "surface_type"],
    coords={
        "channel":       ds_dry.channel.values,
        "surface_type":  ds_dry.surface_type.values,
        "channel_label": ("channel", ds_dry.channel_label.values),
    },
    attrs={"units": "K", "description": "Tmax + Ge-window corrected BT"},
)

# Delta BT due to corrections
delta_BT = BT_corrected - BT_raw
print("Raw-signal effect of filter + window (effective − scene BT) in K:")
print(pd.DataFrame(delta_BT.round(3),
                   index=channels["label"],
                   columns=list(SURFACE_TYPES.keys())))
````

## 9. Sanity checks

Basic physical consistency checks on the simulated brightness temperatures.

````python
print("=== Sanity checks ===")
BT_dry_uu = ds_dry.uu.isel(altitude=0)

# 1. In a dry Arctic atmosphere the window BT should be close to the surface T.
#    A small exceedance (≤ 3 K) would be plausible below a temperature inversion.
ch3_snow = float(BT_dry_uu.sel(channel=3, surface_type="snow_ice"))
T_snow   = SURFACE_TYPES["snow_ice"]["T_sfc"]
assert ch3_snow < T_snow + 3.0, (
    f"CH3 BT ({ch3_snow:.1f} K) should be within 3 K of surface T ({T_snow:.1f} K) in dry Arctic"
)
print(f"✓ CH3 BT ({ch3_snow:.1f} K) within 3 K of surface T ({T_snow:.1f} K)")

# 2. Open water should appear warmer than snow-covered ice in all channels
for ch_id in [3, 5, 6]:
    BT_water = float(BT_dry_uu.sel(channel=ch_id, surface_type="open_water"))
    BT_snow  = float(BT_dry_uu.sel(channel=ch_id, surface_type="snow_ice"))
    assert BT_water > BT_snow, (
        f"CH{ch_id}: open water ({BT_water:.1f} K) should be warmer than snow ice ({BT_snow:.1f} K)"
    )
print(f"✓ Open water BT > snow-ice BT in CH3, CH5, CH6")

# 3. More H₂O changes the 12 µm BT: the surface signal is attenuated and
#    replaced by emission from the water vapour at its own temperature
ch6_dry_snow   = float(ds_dry.uu.isel(altitude=0).sel(channel=6, surface_type="snow_ice"))
ch6_moist_snow = float(ds_moist.uu.isel(altitude=0).sel(channel=6, surface_type="snow_ice"))
print(f"CH6 snow_ice BT: dry={ch6_dry_snow:.2f} K, moist={ch6_moist_snow:.2f} K  "
      f"(Δ = {ch6_moist_snow - ch6_dry_snow:.2f} K)")

# 4. CH1 and CH4 are the same broadband filter and must give the same BT
for stype in list(SURFACE_TYPES.keys()):
    BT_1 = float(BT_dry_uu.sel(channel=1, surface_type=stype))
    BT_4 = float(BT_dry_uu.sel(channel=4, surface_type=stype))
    assert abs(BT_1 - BT_4) < 0.1, (
        f"CH1 and CH4 are both broadband; BT should be identical but "
        f"CH1={BT_1:.2f} K, CH4={BT_4:.2f} K for {stype}"
    )
print(f"✓ CH1 and CH4 (both broadband 7700-12000 nm) yield identical BT")

print()
print("All sanity checks passed.")
````

## 10. Scientific interpretation

### What the channel differences tell us

1. **Split-window (CH3−CH6, CH5−CH6).** With 5 mm of water vapour the CH3 − CH6 difference stays within ±0.35 K for all three surfaces. With 20 mm it grows to 1.7 K over open water and 0.8 K over thin ice. Over the 253 K snow it is −0.3 K: the air above is warmer than the surface, so the more absorbing 12 µm channel sees warmer emission. The sign and size depend on the water-vapour column and on the temperature contrast between surface and air.

2. **Emissivity contrast.** Snow-covered ice (ε ≈ 0.99) and open water (ε ≈ 0.98) have similar broadband emissivities, but their spectral shape differs slightly at 11–12 µm. The spectrally flat emissivity used here underestimates this contrast.

3. **CH2 (8.65 µm).** This channel lies in the window, short of the 9.6 µm ozone band, but has stronger water-vapour absorption than 10.7 µm. Ice and liquid absorb differently between 8.5 and 11 µm, so the CH2 − CH3 difference is the classic cloud-phase indicator; using it requires careful forward modelling.

4. **Cloud masking.** The 1 km thick cloud of section 8 (τ ≈ 12) replaces the surface signal with emission at about the cloud-top temperature: the channels drop by about 16–17 K over open water and 9–10 K over thin ice, but change by less than 0.5 K over the snow, whose temperature is close to that of the cloud top. A window channel alone cannot separate a cloud from a surface of the same temperature.

### What this model does *not* include

- Spectrally resolved SRFs (top-hat is an approximation)
- Measured surface emissivity spectra
- Horizontal inhomogeneity of the scene
- Sensor calibration (offset, non-linearity, shading)
- Validation against VELOX radiance data

## 11. Where to go next

1. **Real SRFs.** Replace `load_srf_or_tophat()` with measured VELOX SRF files and integrate $L_\lambda(\lambda) \cdot R_{\mathrm{ch}}(\lambda)$ over each channel ({doc}`/guide/outputs` describes instrument channels in pyRadtran).

2. **Spectral emissivity.** Load laboratory spectra (ASTER, ECOSTRESS, or Salisbury & D'Aria 1994) with `load_surface_emissivity_csv()`. A wavelength-resolved emissivity changes all channel BTs, the split-window differences most of all.

3. **Real atmosphere.** Replace the US standard atmosphere with ERA5 or CARRA profiles via `era5_atmosphere=` ({doc}`era5_atmosphere`) or with radiosonde profiles from the campaign ({doc}`get_radiosonde`).

4. **Cloud profiles.** Replace the homogeneous cloud layer with profiles reconstructed from lidar and radar (e.g. Cloudnet); see {doc}`/guide/clouds` and {doc}`arctic_cloud_experiment`.

5. **Retrieval.** Use this forward model as the observation operator in an optimal estimation or Bayesian retrieval of surface skin temperature, emissivity or H₂O column. The full VELOX non-uniformity and limb-correction chain needs the `local_delta` and `cos_vza` fields of the zarr file.

6. **Validation.** Compare simulated BTs with the VELOX and KT-19 observations of RF04 or later flights. The KT-19 covers 9.60–11.50 µm, comparable to CH3 (10.74 µm) but not identical.
